# Matched corruption-severity comparisons
Run after generating actual pilot results. These cells never fabricate missing data; they validate artifacts before analysis.


```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
from neuralshift.utils.validation import validate_results
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RESULTS = ROOT / 'results' / 'pilot'
validate_results(RESULTS)
runs = sorted((RESULTS / 'runs').glob('*'))
```

```python
metrics = pd.concat([pd.read_csv(r/'metrics.csv') for r in runs])
raw = metrics.query("method == 'msp' and calibration == 'raw'")
raw.groupby(['corruption','severity']).accuracy.agg(['mean','std','count'])
```

```python
raw.pivot_table(index='severity', columns='corruption', values='accuracy').plot(ylim=(0,1), ylabel='Accuracy', xlabel='Severity')
```
